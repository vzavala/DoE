Bayesian estimation for $Y=\theta X+\epsilon$ with one observation $(x=1,y=2)$
(Victor Z and Jaron T, UW-Madison, 2024).

| Slides | Python | Meaning |
|---|---|---|
| $\theta\sim\mathcal{N}(\bar{\theta},\Sigma_\theta)$ | `theta_bar`, `Sigma_theta` | prior (0, 1) |
| $y\sim\mathcal{N}(\theta x,\sigma^2)$ | `sigma2` | likelihood noise variance (1) |
| $\log f(\theta)$, $\log f(y\mid\theta)$, $\log f(\theta\mid y)$ | `log_f_theta`, `log_f_y_theta`, `log_f_theta_y` | log prior, log likelihood, log posterior (up to a constant) |

**Note:** This notebook reproduces Example 5.15 (Section 5.5) of *Statistics for Chemical Engineers*: prior, likelihood, and posterior of a parameter theta.

**The concept.** Bayes' rule combines what we knew before (prior) with what the data say (likelihood):

$$f(\theta\mid y)=\frac{f(y\mid\theta)\,f(\theta)}{f(y)}\;\propto\;f(y\mid\theta)\,f(\theta)\quad\Rightarrow\quad\log f(\theta\mid y)=\log f(y\mid\theta)+\log f(\theta)+\mathrm{const}.$$

For Gaussian prior and noise the logarithms are parabolas,

$$\log f(\theta)=-\frac{(\theta-\bar{\theta})^2}{2\Sigma_\theta}+\mathrm{const},\qquad \log f(y\mid\theta)=-\frac{(y-\theta x)^2}{2\sigma^2}+\mathrm{const}.$$

Maximizing the likelihood alone gives the MLE. Maximizing the posterior adds the prior as a regularizer.

```python
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(-3, 3, 400)
theta_bar, Sigma_theta = 0.0, 1.0
x, y, sigma2 = 1.0, 2.0, 1.0

log_f_theta = -0.5*np.log(2*np.pi*Sigma_theta) - (theta - theta_bar)**2 / (2*Sigma_theta)
log_f_y_theta = -0.5*np.log(2*np.pi*sigma2) - (y - theta*x)**2 / (2*sigma2)
log_f_theta_y = log_f_y_theta + log_f_theta          # Bayes: f(theta|y) ∝ f(y|theta) f(theta)

fig, axs = plt.subplots(2, 3, figsize=(13, 7))
for k, (vals, label) in enumerate(((log_f_theta, r'f(\theta)'), (log_f_y_theta, r'f(y|\theta)'),
                                   (log_f_theta_y, r'f(\theta|y)'))):
    axs[0, k].plot(theta, vals, 'k', linewidth=1.5)
    axs[0, k].set_ylabel(rf'$\log {label}$')
    axs[1, k].plot(theta, np.exp(vals), 'k', linewidth=1.5)
    axs[1, k].set_ylabel(rf'${label}$')
    for ax in axs[:, k]:
        ax.set_xlabel(r'$\theta$')
        ax.grid(True)
plt.tight_layout()
plt.show()

for vals, name in ((log_f_theta, 'prior'), (log_f_y_theta, 'likelihood (MLE)'), (log_f_theta_y, 'posterior (Bayes)')):
    print(f"{name:18s} is maximized at theta = {theta[np.argmax(vals)]:.2f}")
```

**The math in closed form.** The sum of two parabolas is a parabola, so the posterior is Gaussian. Its variance and mean are

$$\Sigma_{\theta|y}=\left(\frac{x^2}{\sigma^2}+\frac{1}{\Sigma_\theta}\right)^{-1},\qquad \hat{\theta}=\Sigma_{\theta|y}\left(\frac{x\,y}{\sigma^2}+\frac{\bar{\theta}}{\Sigma_\theta}\right).$$

Information adds: the posterior precision is the precision of the data plus the precision of the prior.

```python
Sigma_post = 1 / (x**2 / sigma2 + 1 / Sigma_theta)
theta_post = Sigma_post * (x * y / sigma2 + theta_bar / Sigma_theta)
print(f"posterior: mean = {theta_post:.2f}, variance = {Sigma_post:.2f}")
print(f"MLE (no prior): {x * y / x**2:.2f}")
```

**Note:** As in the book, the posterior blends the prior (old information) and the likelihood (new data): its most likely value, theta = 1, is a trade-off between the two.
